```python
# ============================================================
# MULTI-UTILITY TOOLKIT - SINGLE CELL PROGRAM
# ============================================================
import datetime
import math
import random
import string
import uuid
import time
import secrets
import importlib

# ------------------------------------------------------------
# FILE OPERATIONS
# ------------------------------------------------------------

def create_file(filename):
    with open(filename, "w", encoding="utf-8"):
        pass
    print("File created successfully!")


def write_file(filename, data):
    with open(filename, "w", encoding="utf-8") as file:
        file.write(data)
    print("Data written successfully!")


def read_file(filename):
    with open(filename, "r", encoding="utf-8") as file:
        print("File Content:")
        print(file.read())


def append_file(filename, data):
    with open(filename, "a", encoding="utf-8") as file:
        file.write(data)
    print("Data appended successfully!")

# ------------------------------------------------------------
# MATHEMATICAL OPERATIONS
# ------------------------------------------------------------

def factorial():
    n = int(input("Enter a number: "))

    if n < 0:
        print("Factorial is not defined for negative numbers.")
    else:
        print("Factorial:", math.factorial(n))

def compound_interest():
    p = float(input("Enter principal amount: "))
    r = float(input("Enter rate of interest (in %): "))
    t = float(input("Enter time (in years): "))

    amount = p * (1 + r / 100) ** t
    interest = amount - p

    print(f"Compound Interest: {amount:.2f}")
    print(f"Interest Earned: {interest:.2f}")


def trigonometric():
    angle = float(input("Enter angle in degrees: "))

    radian = math.radians(angle)

    print("sin(", angle, ") =", round(math.sin(radian), 6))
    print("cos(", angle, ") =", round(math.cos(radian), 6))
    print("tan(", angle, ") =", round(math.tan(radian), 6))


def geometric_area():
    print("\n1. Circle")
    print("2. Rectangle")
    print("3. Triangle")
    print("4. Square")

    choice = input("Choose shape: ")

    if choice == "1":
        radius = float(input("Enter radius: "))
        area = math.pi * radius ** 2
        print("Area of Circle:", area)

    elif choice == "2":
        length = float(input("Enter length: "))
        width = float(input("Enter width: "))
        print("Area of Rectangle:", length * width)

    elif choice == "3":
        base = float(input("Enter base: "))
        height = float(input("Enter height: "))
        print("Area of Triangle:", 0.5 * base * height)

    elif choice == "4":
        side = float(input("Enter side: "))
        print("Area of Square:", side ** 2)

    else:
        print("Invalid choice!")


# ------------------------------------------------------------
# RANDOM DATA GENERATION
# ------------------------------------------------------------

def random_number():
    start = int(input("Enter minimum value: "))
    end = int(input("Enter maximum value: "))

    print("Random Number:", random.randint(start, end))


def random_list():
    count = int(input("Enter list size: "))
    start = int(input("Enter minimum value: "))
    end = int(input("Enter maximum value: "))

    data = [random.randint(start, end) for i in range(count)]

    print("Random List:", data)


def random_password():
    length = int(input("Enter password length: "))

    if length < 4:
        print("Password length must be at least 4.")
        return

    characters = string.ascii_letters + string.digits + string.punctuation

    password = "".join(
        secrets.choice(characters)
        for i in range(length)
    )

    print("Generated Password:", password)


def random_otp():
    otp = "".join(
        secrets.choice(string.digits)
        for i in range(6)
    )

    print("Generated OTP:", otp)


# ------------------------------------------------------------
# DATETIME AND TIME OPERATIONS
# ------------------------------------------------------------

def datetime_menu():

    while True:

        print("\n" + "=" * 35)
        print("Datetime and Time Operations")
        print("=" * 35)

        print("1. Display current date and time")
        print("2. Calculate difference between two dates/times")
        print("3. Format date into custom format")
        print("4. Stopwatch")
        print("5. Countdown Timer")
        print("6. Back to Main Menu")

        choice = input("Enter your choice: ")

        # Current Date and Time
        if choice == "1":

            current = datetime.datetime.now()

            print(
                "Current Date and Time:",
                current.strftime("%Y-%m-%d %H:%M:%S")
            )

        # Difference Between Dates
        elif choice == "2":

            d1 = input(
                "Enter the first date (YYYY-MM-DD): "
            )

            d2 = input(
                "Enter the second date (YYYY-MM-DD): "
            )

            date1 = datetime.datetime.strptime(
                d1,
                "%Y-%m-%d"
            )

            date2 = datetime.datetime.strptime(
                d2,
                "%Y-%m-%d"
            )

            difference = abs(
                (date2 - date1).days
            )

            print("Difference:", difference, "days")

        # Custom Format
        elif choice == "3":

            value = input(
                "Enter date/time (YYYY-MM-DD HH:MM:SS): "
            )

            format_value = input(
                "Enter custom format "
                "(e.g. %d-%m-%Y %I:%M %p): "
            )

            dt = datetime.datetime.strptime(
                value,
                "%Y-%m-%d %H:%M:%S"
            )

            print(
                "Formatted Date:",
                dt.strftime(format_value)
            )

        # Stopwatch
        elif choice == "4":

            print("Stopwatch started.")
            print("Press Enter to stop.")

            start = time.time()

            input()

            end = time.time()

            elapsed = end - start

            print(
                "Elapsed time:",
                round(elapsed, 2),
                "seconds"
            )

        # Countdown
        elif choice == "5":

            seconds = int(
                input("Enter countdown time in seconds: ")
            )

            while seconds > 0:

                print(
                    f"\rTime remaining: "
                    f"{seconds:02d} seconds",
                    end=""
                )

                time.sleep(1)

                seconds -= 1

            print("\nCountdown finished!")

        elif choice == "6":
            break

        else:
            print("Invalid choice!")

        input("\nPress Enter to continue...")

# -----------------------------------------------------------
# MATHEMATICAL MENU
# -----------------------------------------------------------

def mathematical_menu():

    while True:

        print("\n" + "=" * 35)
        print("Mathematical Operations")
        print("=" * 35)

        print("1. Calculate Factorial")
        print("2. Solve Compound Interest")
        print("3. Trigonometric Calculations")
        print("4. Area of Geometric Shapes")
        print("5. Back to Main Menu")

        choice = input("Enter your choice: ")

        try:

            if choice == "1":
                factorial()

            elif choice == "2":
                compound_interest()

            elif choice == "3":
                trigonometric()

            elif choice == "4":
                geometric_area()

            elif choice == "5":
                break

            else:
                print("Invalid choice!")

        except ValueError:
            print("Please enter a valid number.")

        if choice != "5":
            input("\nPress Enter to continue...")


# ------------------------------------------------------------
# RANDOM MENU
# ------------------------------------------------------------

def random_menu():

    while True:

        print("\n" + "=" * 35)
        print("Random Data Generation")
        print("=" * 35)

        print("1. Generate Random Number")
        print("2. Generate Random List")
        print("3. Create Random Password")
        print("4. Generate Random OTP")
        print("5. Back to Main Menu")

        choice = input("Enter your choice: ")

        try:

            if choice == "1":
                random_number()

            elif choice == "2":
                random_list()

            elif choice == "3":
                random_password()

            elif choice == "4":
                random_otp()

            elif choice == "5":
                break

            else:
                print("Invalid choice!")

        except ValueError:
            print("Please enter valid numbers.")

        if choice != "5":
            input("\nPress Enter to continue...")


# ------------------------------------------------------------
# UUID
# ------------------------------------------------------------

def generate_uuid():

    print("\n" + "=" * 35)
    print("Generate Unique Identifiers (UUID)")
    print("=" * 35)

    unique_id = uuid.uuid4()

    print("Generated UUID:", unique_id)

    input("\nPress Enter to continue...")


# ------------------------------------------------------------
# FILE MENU
# ------------------------------------------------------------

def file_menu():

    while True:

        print("\n" + "=" * 35)
        print("File Operations (Custom Module)")
        print("=" * 35)

        print("1. Create a new file")
        print("2. Write to a file")
        print("3. Read from a file")
        print("4. Append to a file")
        print("5. Back to Main Menu")

        choice = input("Enter your choice: ")

        try:

            if choice == "1":

                filename = input(
                    "Enter file name: "
                )

                create_file(filename)

            elif choice == "2":

                filename = input(
                    "Enter file name: "
                )

                data = input(
                    "Enter data to write: "
                )

                write_file(filename, data)

            elif choice == "3":

                filename = input(
                    "Enter file name: "
                )

                read_file(filename)

            elif choice == "4":

                filename = input(
                    "Enter file name: "
                )

                data = input(
                    "Enter data to append: "
                )

                append_file(filename, data)

            elif choice == "5":
                break

            else:
                print("Invalid choice!")

        except FileNotFoundError:
            print("File not found!")

        except OSError as e:
            print("File error:", e)

        if choice != "5":
            input("\nPress Enter to continue...")


# ------------------------------------------------------------
# DIR() MODULE EXPLORATION
# ------------------------------------------------------------

def explore_module():

    print("\n" + "=" * 35)
    print("Explore Module Attributes (dir())")
    print("=" * 35)

    module_name = input(
        "Enter module name to explore: "
    )

    try:

        module = importlib.import_module(
            module_name
        )

        attributes = [
            item for item in dir(module)
            if not item.startswith("__")
        ]

        print(
            f"\nAvailable Attributes in "
            f"{module_name} module:"
        )

        print(attributes)

    except ModuleNotFoundError:

        print(
            "Module not found!"
        )

    input("\nPress Enter to continue...")


# ------------------------------------------------------------
# MAIN MENU
# ------------------------------------------------------------
def main():

    while True:

        print("\n" + "=" * 40)
        print("Welcome to Multi-Utility Toolkit")
        print("=" * 40)

        print("Choose an option:")

        print("1. Datetime and Time Operations")
        print("2. Mathematical Operations")
        print("3. Random Data Generation")
        print("4. Generate Unique Identifiers (UUID)")
        print("5. File Operations (Custom Module)")
        print("6. Explore Module Attributes (dir())")
        print("7. Exit")

        print("=" * 40)

        choice = input(
            "Enter your choice: "
        )

        if choice == "1":
            datetime_menu()

        elif choice == "2":
            mathematical_menu()

        elif choice == "3":
            random_menu()

        elif choice == "4":
            generate_uuid()

        elif choice == "5":
            file_menu()

        elif choice == "6":
            explore_module()

        elif choice == "7":

            print("\n" + "=" * 35)
            print(
                "Thank you for using the "
                "Multi-Utility Toolkit!"
            )
            print("=" * 35)

            break

        else:
            print(
                "Invalid choice. "
                "Please try again."
            )


# ============================================================
# PROGRAM START
# ============================================================

if __name__ == "__main__":
    main()
```
